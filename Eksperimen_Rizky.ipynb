```python
import pandas as pd
import numpy as np

print("Loading data...")
df = pd.read_csv('../stroke_raw/healthcare-dataset-stroke-data.csv')
display(df.head())

df_clean = df.drop(columns=['id'])

df_clean['bmi'] = pd.to_numeric(df_clean['bmi'], errors='coerce') 
df_clean['bmi'] = df_clean['bmi'].fillna(df_clean['bmi'].median()) 

df_clean = df_clean[df_clean['gender'] != 'Other']

df_encoded = pd.get_dummies(df_clean, drop_first=True)

print("\nInfo data setelah dibersihkan:")
display(df_encoded.info())

df_encoded.to_csv('stroke_preprocessing/dataset_bersih.csv', index=False)
print("\nData berhasil diproses dan disimpan di folder stroke_preprocessing!")
```
